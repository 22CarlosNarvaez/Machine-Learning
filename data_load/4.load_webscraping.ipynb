```python
# IMPORTAMOS LIBRERIAS

import requests
import pandas as pd

from bs4 import BeautifulSoup
```

```python
#CONEXION A LA PAGINA WEB

url = "https://books.toscrape.com/"

response = requests.get(url)

print(response.status_code)
```

```python
#REVISAMOS EL CONTENIDO HTML

print(response.text[:1000])
```

```python
#CREAMOS EL OBJETO BEAUTIFULSOUP

soup = BeautifulSoup(response.text, "html.parser")

print(soup.prettify()[:2000])
```

```python
#BUSCAMOS TODOS LOS CONTENEDORES DE LIBROS  

libros = soup.find_all("article", class_="product_pod")

print("Total de libros encontrados:", len(libros))
```

```python
# EXTRAEMOS EL NOMBRE DE LOS LIBROS

nombres = []

for libro in libros:

    nombre = libro.h3.a["title"]

    nombres.append(nombre)

print(nombres)
```

```python
#EXTRAER PRECIOS DE LOS LIBROS Y OTROS CAMPOS

datos = []

for libro in libros:

    nombre = libro.h3.a["title"]

    precio = libro.find(
        "p", class_="price_color"
    ).text

    disponibilidad = libro.find(
        "p", class_="instock availability"
    ).text.strip()

    calificacion = libro.find(
        "p", class_="star-rating"
    )["class"][1]

    datos.append({
        "nombre": nombre,
        "precio": precio,
        "disponibilidad": disponibilidad,
        "calificacion": calificacion
    })

print(datos)
```

```python
#CONVERTIR LOS DATOS A UN DATAFRAME DE PANDAS

df_libros = pd.DataFrame(datos)

df_libros.head(10)
```

```python
#CONVERTIR LAS CALIFICACIONES EN NUMERO 

mapa_calificaciones = {
    "One": 1,
    "Two": 2,
    "Three": 3,
    "Four": 4,
    "Five": 5
}

df_libros["calificacion"] = (
    df_libros["calificacion"]
    .map(mapa_calificaciones)
)
```
