```python
def obtener_todos(endpoint):
    
    url = f"https://rickandmortyapi.com/api/{endpoint}"

    resultados = []

    while url:

        response = requests.get(url, timeout=30)
        response.raise_for_status()

        data = response.json()

        resultados.extend(data["results"])

        url = data["info"]["next"]

    return pd.DataFrame(resultados)

df_personajes = obtener_todos("character")

df_location = obtener_todos("location")

df_episode = obtener_todos("episode")

print("PERSONAJES:", df_personajes.shape)

print("UBICACIONES:", df_location.shape)

print("EPISODIOS:", df_episode.shape)
```
