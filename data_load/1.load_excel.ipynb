What is Pandas?
Pandas is a Python library used for working with data sets.

It has functions for analyzing, cleaning, exploring, and manipulating data.

The name "Pandas" has a reference to both "Panel Data", and "Python Data Analysis" and was created by Wes McKinney in 2008.

Why Use Pandas?

Pandas can clean messy data sets, and make them readable and relevant.

Relevant data is very important in data science.

What Can Pandas Do?

Is there a correlation between two or more columns?
What is average value?
Max value?
Min value?

## MANEJO DE DATOS FALTANTES Y DUPLICADOS


```python
#1. Identificar valores nulos en el dataset
df.isnull().sum() #display the number of null values in each column of the
```

```python
# 2. ELiminar filas con valores nulos
df.dropna()
```

```python
# 3. Rellenar valores nulos con algo en especifico

df.fillna(0)
```

```python
# 4. Rellenar valores nulos con la media de la columna
df['precio_unitario'] = df['precio_unitario'].fillna(df['precio_unitario'].mean())
df.head()
```

## 1. Averiguar como general ver los datos duplicados

```python
# 5. Verificar si hay valores decimales duplicados en el dataset
# Cambia 'nombre_columna' por el nombre exacto de tu columna
df.duplicated(subset=['precio_unitario']).sum()
 #display the number of duplicated rows in the dataframe
```

## 2. Codifique variables categoricas usando LabelEncoder y One-Hot Encoding

```python
# 6. Convertir la columna 'precio_unitario' a valores numéricos
#LabelEncoder
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['categoria_num'] = le.fit_transform(df['categoria'])
#One-Hot Encoding
df_ciudad = pd.get_dummies(df, columns=['ciudad'], dtype=int)    
```

## 3. Aplique normalizacion y estandarización

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

columnas_numericas = df.select_dtypes(include=['int64', 'float64']).columns
datos_num = df[columnas_numericas]

#normalizar los datos numéricos 

normalizador = MinMaxScaler()
datos_normalizados = normalizador.fit_transform(datos_num)

df_normalizado = pd.DataFrame(datos_normalizados, columns=columnas_numericas)
print("Datos Normalizados:")
display(df_normalizado.head())

#aplicar estandarización a los datos numéricos
estandarizador = StandardScaler()
datos_estandarizados = estandarizador.fit_transform(datos_num)
print("Datos Estandarizados: ")
df_estandarizado = pd.DataFrame(datos_estandarizados, columns=columnas_numericas)
display(df_estandarizado.head())
```

## 4. Balancee la variable objetivo usando SMOTE

```python
#balancear el dataset utilizando SMOTE

from imblearn.over_sampling import SMOTE

# 1. Definir X (solo columnas numéricas) e y
X = df.drop('ciudad', axis=1).select_dtypes(include=['int64', 'float64'])
y = df['ciudad']

# 2. Aplicar SMOTE usando X mayúscula
smote = SMOTE(random_state=42)
X_balanced, y_balanced = smote.fit_resample(X, y)
```

## 5. Muestre el conjunto de datos final para entrenar un modelo de clasificación


```python
#mostramos el conjunto de datos final para entrenar el modelo de machine learning de clasificación

df_X_final = pd.DataFrame(X_balanced, columns=X.columns)

df_final_entrenamiento = df_X_final.copy()
df_final_entrenamiento['ciudad'] = y_balanced.values

print(f"Dimensiones del dataset final: {df_final_entrenamiento.shape[0]} filas y {df_final_entrenamiento.shape[1]} columnas")
print("\nDistribución final de clases:")
print(df_final_entrenamiento['ciudad'].value_counts())

print("\nVista previa del conjunto de datos final:")
display(df_final_entrenamiento.head(10))


```
